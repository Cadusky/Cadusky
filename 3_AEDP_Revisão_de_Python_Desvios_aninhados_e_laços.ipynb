2 - Faça um programa em Python que solicite um código referente ao tipo da diária de hospedagem e também a quantidade de diárias desejada por um cliente. Calcule e mostre, usando estrutura condicional aninhada, o valor total a pagar pelo cliente, conforme a tabela abaixo:

Imagem2.png

```python
error = True

while True:
  tipo_diaria = str(input("Qual o tipo da diária?\n(S) Simples\n(D) Duplo\n(T) Triplo\n:").upper())

  while error:
    try:
      dias = int(input("Quantos dias de hospedagem?\n:"))
      error = False
    except ValueError:
      print("Informação inválida tente novamente")

  if tipo_diaria == "S":
    valor = 255.50
    break
  elif tipo_diaria == "D":
    valor = 305.50
    break
  elif tipo_diaria == "T":
    valor = 360.50
    break
  else:
    print("Opção inválida insira novamente: ")

print("O valor da hospedagem é de R${:.2f}".format(valor * dias))
```

4- Na cidade de São Paulo a circulação de veículos é restrita a determinados horários no chamado “Centro Expandido”. Com base no digito final da placa, o motorista sabe se deve ou não circular no horário proibido. Considere a tabela abaixo:

Imagem4.png

Faça um programa que receba o número inteiro da placa de um automóvel, verifique e imprima qual o dia da semana o veículo não pode circular no “Centro Expandido”


```python
while True:
  placa = str(input("Qual a numeração da placa? "))
  final = placa[-1]
  if final == "1" or final == "2":
    nao_circular = "Segunda-Feira"
    break
  elif final == "3" or final == "4":
    nao_circular = "Terça-Feira"
    break
  elif final == "5" or final == "6":
    nao_circular = "Quarta-Feira"
    break
  elif final == "7" or final == "8":
    nao_circular = "Quinta-Feira"
    break
  elif final == "9" or final == "0":
    nao_circular = "Sexta-Feira"
    break
  else:
    print("Placa ínvalida tente novamente")
print("o veículo não pode circular no seguinte dia da semana: {}".format(nao_circular))
```

5- Crie um algoritmo que leia a idade de uma pessoa e informe sua classe eleitoral:

*	não-eleitor (abaixo de 16 anos)
*	eleitor obrigatório (entre 18 e 65 anos)
*	eleitor facultativo (entre 16 e 18 anos e maior de 65 anos)



```python
error = True
while error == True:
  try:
    idade = int(input("Digite sua idade: "))
    error = False
  except ValueError:
    print("Valor inválido tente novamente")
if idade >= 16 and idade < 18 or idade > 65:
  estado = "eleitor facultativo"
else:
  if idade < 16:
    estado = "não-eleitor"
  elif idade >= 18 and idade <= 65:
    estado = "eleitor obrigatório"
print(estado)
```

7- Faça um programa em Python que solicite ao usuário 10 números reais, calcule e mostre a soma dos números digitados.

```python
soma = 0
for i in range(1, 11):
  num = int(input("Digite o número {}: ".format(i)))
  soma += num
print(soma)
```

8- Faça um programa em Python que solicite ao usuário uma quantidade qualquer de números reais, calcule e mostre a média dos números digitados.

```python
soma = 0.0
cont = 0
continua = True

while continua:
  try:
    soma += float(input("Digite o número {} (S para sair): ".format(cont + 1)))
    cont += 1
  except ValueError:
    continua = False

if cont > 0:
  print("A média dos número é {}".format(soma / cont))
else:
  print(("Não foram fornecidos números."))
```
